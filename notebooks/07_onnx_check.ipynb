## Part 7: does the small served retriever lose anything?

The service runs the retriever with ONNX Runtime on a CPU, with int8 weights (278 MB instead of 1.1 GB) and inputs cut at 256 tokens. This notebook measures what that costs in retrieval, on the full corpus and the test and Najiz questions, against the PyTorch model used in parts 1 to 4.

The first run measured the int8 model from part 5 and found it lost 13 points of Recall@1 on Najiz (63.3% → 50.6%) and ran slower on Kaggle's CPU. This version exports again: float32, and int8 for the matrix multiplications only, with one scale per output channel.

**Settings:** Internet on, no accelerator (the service runs on CPU).
